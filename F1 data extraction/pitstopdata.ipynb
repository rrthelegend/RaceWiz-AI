```python
import fastf1
import pandas as pd
fastf1.Cache.enable_cache('/Users/rahulrajagopal/Documents/PBL - RaceWiz AI')
```

```python
pit_stop_data = []

for year in range(2018, 2024):
    for round_number in range(1, 22):
        try:
            session = fastf1.get_session(year, round_number, 'R')
            session.load()
            pit_stops = session.laps[session.laps['PitOutTime'].notna()][['Driver', 'LapNumber', 'PitOutTime', 'Compound']]
            pit_stops['Year'] = year
            pit_stops['Round'] = round_number
            pit_stop_data.append(pit_stops)
        except Exception as e:
            print(f"Failed to load data for {year} Round {round_number}: {e}")

pit_stops_df = pd.concat(pit_stop_data, ignore_index=True)
print(pit_stops_df.head())
pit_stop_data.to_csv('pit_stop_data.csv', index=False)
```
