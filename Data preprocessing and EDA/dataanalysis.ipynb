```python
cache_dir = '/Users/rahulrajagopal/Documents/PBL - RaceWiz AI/F1 data extraction/race_data'
fastf1.Cache.enable_cache(cache_dir)

all_lap_data = []
weather_data = []
race_events = []

for year in range(2018, 2025): 
    for round_number in range(1, 23):  
        try:
            session_path = os.path.join(cache_dir, f"{year}-{round_number}-R.ff1pkl")
            if os.path.exists(session_path):
                session = fastf1.api.Cache.load(session_path)

                laps = session.laps[['Driver', 'LapNumber', 'LapTime', 'Compound', 'Sector1Time', 'Sector2Time', 'Sector3Time', 'PitInTime', 'PitOutTime']]
                laps['Year'] = year
                laps['Round'] = round_number
                all_lap_data.append(laps)

                weather = pd.DataFrame(session.weather_data)
                weather['Year'] = year
                weather['Round'] = round_number
                weather_data.append(weather)

                events = pd.DataFrame(session.events)
                events['Year'] = year
                events['Round'] = round_number
                race_events.append(events)
            else:
                print(f"No cached data found for {year} Round {round_number}")

        except Exception as e:
            print(f"Failed to load data for {year} Round {round_number}: {e}")

lap_data_df = pd.concat(all_lap_data, ignore_index=True)

weather_data_df = pd.concat(weather_data, ignore_index=True)

race_events_df = pd.concat(race_events, ignore_index=True)

lap_data_df.to_csv('lap_data_2018_2024.csv', index=False)
weather_data_df.to_csv('weather_data_2018_2024.csv', index=False)
race_events_df.to_csv('race_events_2018_2024.csv', index=False)
```

```python

all_lap_data = []
all_weather_data = []
all_race_events = []

for year in range(2018, 2025):
    for round_number in range(1, 23):  
        try:
            session = fastf1.get_session(year, round_number, 'R')
            session.load()  

            laps = session.laps[['Driver', 'LapNumber', 'LapTime', 'Compound', 'Sector1Time', 'Sector2Time', 'Sector3Time', 'PitInTime', 'PitOutTime']]
            laps['Year'] = year
            laps['Round'] = round_number
            all_lap_data.append(laps)

            weather = pd.DataFrame(session.weather_data)
            weather['Year'] = year
            weather['Round'] = round_number
            all_weather_data.append(weather)

            events = pd.DataFrame(session.events)
            events['Year'] = year
            events['Round'] = round_number
            all_race_events.append(events)

            print(f"Successfully processed {year} Round {round_number}")

        except Exception as e:
            print(f"Failed to load {year} Round {round_number}: {e}")

lap_data_df = pd.concat(all_lap_data, ignore_index=True) if all_lap_data else pd.DataFrame()
weather_data_df = pd.concat(all_weather_data, ignore_index=True) if all_weather_data else pd.DataFrame()
race_events_df = pd.concat(all_race_events, ignore_index=True) if all_race_events else pd.DataFrame()

lap_data_df.to_csv('lap_data_2018_2024.csv', index=False)
weather_data_df.to_csv('weather_data_2018_2024.csv', index=False)
race_events_df.to_csv('race_events_2018_2024.csv', index=False)
```

```python
telemetry_data = []

for year in range(2018, 2025):
    for round_number in range(1, 23):  # Adjust for the number of rounds in a season
        try:
            session = fastf1.get_session(year, round_number, 'R')
            session.load()
            
            for driver in session.drivers:
                laps = session.laps.pick_driver(driver)
                for lap in laps.iterlaps():
                    telemetry = lap[1].get_telemetry()
                    telemetry['Year'] = year
                    telemetry['Round'] = round_number
                    telemetry['Driver'] = driver
                    telemetry['LapNumber'] = lap[1]['LapNumber']
                    telemetry_data.append(telemetry)
                    
        except Exception as e:
            print(f"Failed to load telemetry for {year} Round {round_number}: {e}")

telemetry_df = pd.concat(telemetry_data, ignore_index=True)
telemetry_df.to_csv('telemetry_data_2018_2024.csv', index=False)
```

```python
pit_stop_data = []

for year in range(2018, 2025):
    for round_number in range(1, 23):
        try:
            session = fastf1.get_session(year, round_number, 'R')
            session.load()
            pit_data = session.get_pitstops()
            
            if not pit_data.empty:
                pit_data['Year'] = year
                pit_data['Round'] = round_number
                pit_stop_data.append(pit_data)
        
        except Exception as e:
            print(f"Failed to load pit stop data for {year} Round {round_number}: {e}")

if pit_stop_data:
    pit_stop_df = pd.concat(pit_stop_data, ignore_index=True)
    pit_stop_df.to_csv('pit_stop_data_2018_2024.csv', index=False)
else:
    print("No pit stop data found for any session.")
```
